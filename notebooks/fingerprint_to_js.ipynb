```python
def tensor_to_js(arr, name):
    lines = []
    lines.append(f'const {name} = [')

    for channel in arr:
        lines.append(' '*2 + '[')
        for row in channel:
            lines.append(' '*4 + '[' + r', '.join(fr'{x}' for x in row) + '],')
        lines.append(' '*2 + '],')

    lines.append('];')

    lines.append('')
    lines.append(f'export default {name};')
    return '\n'.join(lines)
```

```python
import torch

fp = torch.load('../ppt_assets/intro/fingerprints/waldek/fp6.pth').float()[None, ...]
```

```python
from pathlib import Path

Path('../animations/07-fingerprint-zoom/waldek_fp.js').write_text(tensor_to_js(fp, 'waldek_fp'))
```
